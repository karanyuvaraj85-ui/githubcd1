```python
from itertools import combinations
import pandas as pd
import numpy as np

# Load the dataset from a CSV file
try:
    dataset = pd.read_csv(r"C:\Users\Student\Downloads\archive (3)\CC GENERAL.csv")
except FileNotFoundError:
    # Fallback to a synthetic dataset with numeric columns if the local file is not found
    print("Local dataset not found. Using fallback synthetic credit card dataset.")
    dataset = pd.DataFrame({
        'BALANCE': np.random.uniform(10, 5000, 100),
        'PURCHASES': np.random.uniform(0, 3000, 100),
        'ONEOFF_PURCHASES': np.random.uniform(0, 2000, 100),
        'INSTALLMENTS_PURCHASES': np.random.uniform(0, 1500, 100),
        'CASH_ADVANCE': np.random.uniform(0, 4000, 100)
    })

# Mapper function to calculate correlation metrics
def mapper(dataset):
    # Get only numerical columns from the dataset
    numerical_columns = dataset.select_dtypes(include=[np.number])
    # Store correlation results for all combinations
    correlation_results = []

    # Calculate Pearson, Spearman, and Kendall correlations for each pair
    for col1, col2 in combinations(numerical_columns.columns, 2):
        pearson_corr = numerical_columns[col1].corr(numerical_columns[col2], method='pearson')
        spearman_corr = numerical_columns[col1].corr(numerical_columns[col2], method='spearman')
        kendall_corr = numerical_columns[col1].corr(numerical_columns[col2], method='kendall')
        # Append the results as a tuple
        correlation_results.append((col1, col2, pearson_corr, spearman_corr, kendall_corr))

    return correlation_results

# Reducer function to simply return the mapped data
def reducer(mapped_data):
    # Since the mapper already computes the correlations, the reducer can just return the results
    return mapped_data

# Step 3: Run MapReduce
# Map step
mapped_data = mapper(dataset)

# Reduce step
correlation_results = reducer(mapped_data)

# Convert the correlation results to a DataFrame for better readability
correlation_df = pd.DataFrame(correlation_results, columns=['Variable 1', 'Variable 2', 'Pearson', 'Spearman', 'Kendall'])

# Display the correlation results
print(correlation_df)
```
