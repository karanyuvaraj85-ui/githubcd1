```python
from collections import defaultdict
from statistics import median, mode, variance, stdev


numbers = [10, 20, 20, 30, 40, 50, 50, 50, 60, 70]


def map_function(number):

    return [("sum", number), ("count", 1), ("number", number)]


def reduce_function(mapped_data):

    total_sum = 0
    total_count = 0
    number_list = []
    frequency = defaultdict(int)


    for key, value in mapped_data:
        if key == "sum":
            total_sum += value
        elif key == "count":
            total_count += 1
        elif key == "number":
            number_list.append(value)
            frequency[value] += 1


    mean_value = total_sum / total_count if total_count != 0 else 0


    median_value = median(number_list) if number_list else None

    mode_value = max(frequency, key=frequency.get) if frequency else None

    variance_value = variance(number_list) if len(number_list) > 1 else 0
    stddev_value = stdev(number_list) if len(number_list) > 1 else 0

    return mean_value, median_value, mode_value, variance_value, stddev_value


for number in numbers:
    mapped_data.extend(map_function(number))


mean_value, median_value, mode_value, variance_value, stddev_value = reduce_function(mapped_data)


print(f"Mean: {mean_value}")
print(f"Median: {median_value}")
print(f"Mode: {mode_value}")
print(f"Variance: {variance_value}")
print(f"Standard Deviation: {stddev_value}")
```
